```python
import os
import numpy as np
import pandas as pd
from tqdm import tqdm
tqdm.pandas()
import shutil
import zipfile
import warnings
warnings.filterwarnings('ignore')


output_dir = f"/content/drive/MyDrive/logo_recognition_similarity_search_project/output"
os.makedirs(output_dir,exist_ok=True)
images_category_dir = f"{output_dir}/images_category"
os.makedirs(images_category_dir,exist_ok=True)

zip_images_dir = f"/content/drive/MyDrive/Data/images_128x128_new.zip"
n_components=600
ds_path = f"{output_dir}/logo-recognition-with-{n_components}-hdbscan-category.tsv"
```



---



```python
df = pd.read_csv(ds_path,sep="\t")
df.head()
```

```python
import zipfile

with zipfile.ZipFile(zip_images_dir, 'r') as zip_ref:
    zip_ref.extractall('/content')
```

```python
df["path_category"] = df.progress_apply(lambda row: os.path.join("images_category",
                                                                 f"category_{row['category']}",
                                                                 os.path.basename(row["path"])
                                                                 ), axis=1)
```

```python
df.head()
```



---



```python
for _, row in tqdm(df.iterrows(), total=len(df), desc="Moving Files"):
    category_folder = os.path.dirname(row["path_category"])
    os.makedirs(category_folder, exist_ok=True)
    path = row["path"]
    path = os.path.join('/content/images_128x128_new/images_128x128',*path.split('/')[1:])
    path_category = row["path_category"]
    shutil.move(path, path_category)
```

```python
category_counts = {cat: len(os.listdir(os.path.join("images_category", cat)))
                   for cat in os.listdir("images_category")}

category_counts
```

```python
from huggingface_hub import HfApi
from google.colab import userdata

HF_TOKEN = userdata.get('HF_TOKEN')
HF_USERNAME = userdata.get('HF_USERNAME')

api = HfApi(token=HF_TOKEN)

repo_id = f"{HF_USERNAME}/images-category"
repo = api.create_repo(repo_id=repo_id, private=True, exist_ok=True, repo_type="dataset")
repo
```

```python
api.upload_file(
    path_or_fileobj="/content/images_category.zip",
    path_in_repo="images_category.zip",
    repo_id=repo_id,
    repo_type="dataset"
    )
```

```python
api.upload_file(
    path_or_fileobj=f"{output_dir}/final-logo-recognition-with-{n_components}-hdbscan-category.tsv",
    path_in_repo=f"final-logo-recognition-with-{n_components}-hdbscan-category.tsv",
    repo_id=repo_id,
    repo_type="dataset"
    )
```
