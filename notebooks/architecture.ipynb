## Abstract

Logo recognition and similarity retrieval are fundamental problems in computer vision, with applications in brand monitoring, counterfeit detection, and content-based image search. Traditional approaches often exhibit limited robustness to real-world variability due to reliance on labeled data and handcrafted features. This study proposes a scalable, annotation-free deep learning pipeline that leverages synthetic logo data, multimodal embedding fusion, unsupervised clustering, and triplet-based metric learning to enable efficient and accurate logo retrieval. A large-scale synthetic dataset comprising 1.77 million logo images paired with descriptive text prompts was utilized to simulate diverse branding scenarios. Visual features were extracted using a frozen ResNet50 model, while textual embeddings were obtained via the MiniLM encoder. These multimodal embeddings were concatenated and reduced in dimensionality using UMAP, then clustered using HDBSCAN to generate over 209,000 pseudo-categories. These clusters served as weak supervision signals for sampling training triplets, enabling the TripletNet model to learn a compact 256-dimensional embedding space. Quantitative evaluation on a held-out test set of over 319,000 logos demonstrated strong retrieval performance. The model achieved a Recall@10 of 98.6% when evaluated using ResNet50-based visual labels and 58.7% using unsupervised category-based labels. These results highlight the model’s capacity to capture both low-level visual and high-level semantic similarity without manual labeling. The proposed framework establishes a robust and generalizable foundation for large-scale logo retrieval and can be extended further with advanced sampling strategies and vision transformer backbones to enhance semantic sensitivity.
